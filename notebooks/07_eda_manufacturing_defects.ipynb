```python
import os, sys
sys.path.append(r'C:/NIRMAAN AI')
import pandas as pd
import numpy as np

csv_path = r'C:/NIRMAAN AI/DATASET/08_MANUFACTURING_DEFECTS/raw/manufacturing_defect_dataset.csv'
df = pd.read_csv(csv_path)
print(f'Shape: {df.shape}')
print('DefectStatus Distribution:', df['DefectStatus'].value_counts().to_dict())
```

## 1. Operational Factors & Financial Costs


```python
print('Mean Operational & Cost Metrics by Defect Status:')
print(df.groupby('DefectStatus')[['ProductionCost', 'DowntimePercentage', 'EnergyConsumption', 'SupplierQuality']].mean())
```

## 2. Key Findings for NirmaanAI Engine
1. Defective runs show a +2.58% higher mean production cost (₹12,473.17 vs ₹12,158.88).
2. DowntimePercentage exhibits high variance (std 1.44%) but near identical means across defect classes (2.49% vs 2.50%).

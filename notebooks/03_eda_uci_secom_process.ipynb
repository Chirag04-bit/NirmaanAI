## 1. High Dimensionality & Missing Value Audit


```python
profile = profile_dataframe(df, target_col='Pass/Fail')
print(f'Total Missing Cells: {profile["total_missing_cells"]:,}')
print(f'Columns with Missing Values: {profile["columns_with_missing"]} / {profile["cols"]}')
print('Target Distribution (-1=Pass, 1=Fail):', profile['target_summary']['distribution'])
```

## 2. Low Variance & Extreme Missingness Screening


```python
missing_pct = df.isnull().mean()
high_missing_cols = missing_pct[missing_pct > 0.40].index.tolist()
numeric_cols = df.select_dtypes(include=[np.number]).columns.drop('Pass/Fail', errors='ignore')
zero_var_cols = [c for c in numeric_cols if df[c].std() == 0 or df[c].nunique() <= 1]
print(f'Columns with >40% missingness: {len(high_missing_cols)}')
print(f'Zero-variance constant columns: {len(zero_var_cols)}')
```

## 3. Key Findings for NirmaanAI Engine
1. Extreme dimensionality (590 features) requires variance thresholding and PCA dimensionality reduction.
2. Fold-isolated imputation is mandatory to prevent train-test data leakage.

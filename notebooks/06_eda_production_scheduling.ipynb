```python
import os, sys
sys.path.append(r'C:/NIRMAAN AI')
import pandas as pd
import numpy as np

csv_path = r'C:/NIRMAAN AI/DATASET/06_MANUFACTURING_PRODUCTION/raw/hybrid_manufacturing_categorical.csv'
df = pd.read_csv(csv_path)
print(f'Shape: {df.shape}')
print('Job Status Distribution:', df['Job_Status'].value_counts().to_dict())
```

## 1. Scheduled vs Actual Delay Analysis


```python
df['Scheduled_Start'] = pd.to_datetime(df['Scheduled_Start'])
df['Actual_Start'] = pd.to_datetime(df['Actual_Start'])
df['Start_Delay_min'] = (df['Actual_Start'] - df['Scheduled_Start']).dt.total_seconds() / 60.0
print('Start Delay (minutes) by Job Status:\n', df.groupby('Job_Status')['Start_Delay_min'].describe())
```

## 2. Key Findings for NirmaanAI Engine
1. Delayed and Failed jobs exhibit initial queue dispatch delays exceeding 10 minutes.
2. Machine availability and operation complexity directly govern bottleneck queue formations in Phase 8.

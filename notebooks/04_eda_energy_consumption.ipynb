```python
import os, sys
sys.path.append(r'C:/NIRMAAN AI')
import pandas as pd
import numpy as np

txt_path = r'C:/NIRMAAN AI/DATASET/04_ENERGY/raw/LD2011_2014.txt'
# Load 1 month sample (2880 15-min intervals) to inspect structure efficiently
df_sample = pd.read_csv(txt_path, sep=';', nrows=2880, decimal=',')
df_sample.rename(columns={df_sample.columns[0]: 'Timestamp'}, inplace=True)
df_sample['Timestamp'] = pd.to_datetime(df_sample['Timestamp'])
print(f'1-Month Profile Shape: {df_sample.shape}')
print(f'Timespan: {df_sample["Timestamp"].min()} to {df_sample["Timestamp"].max()}')
```

## 1. Diurnal Cyclicity & Industrial Load Patterns (Client MT_124)


```python
client_col = 'MT_124' # Active from January 2011
df_sample['Hour'] = df_sample['Timestamp'].dt.hour
hourly_mean = df_sample.groupby('Hour')[client_col].mean()
print(f'Hourly Average Load for {client_col} (kW):\n', hourly_mean.head(12))
```

## 2. Key Findings & Operational Assumptions
1. **Empirical Finding**: 24-hour diurnal cyclicity is prominent across active client profiles.
2. **Configured Operational Assumption**: Peak tariff hours (18:00–22:00 at ₹12.50/kWh) represent an external Indian MSME tariff rule from `factory_defaults.yaml` (not an empirical discovery of the Portuguese grid dataset).

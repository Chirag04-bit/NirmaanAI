```python
import os, sys
sys.path.append(r'C:/NIRMAAN AI')
import pandas as pd
import numpy as np

txt_path = r'C:/NIRMAAN AI/DATASET/04_ENERGY/raw/LD2011_2014.txt'
df_sample = pd.read_csv(txt_path, sep=';', nrows=2880, decimal=',')
df_sample.rename(columns={df_sample.columns[0]: 'Timestamp'}, inplace=True)
df_sample['Timestamp'] = pd.to_datetime(df_sample['Timestamp'])
print(f'1-Month Profile Shape: {df_sample.shape}')
print(f'Timespan: {df_sample["Timestamp"].min()} to {df_sample["Timestamp"].max()}')
```

## 1. Diurnal Cyclicity & Industrial Load Patterns


```python
client_col = df_sample.columns[1]
df_sample['Hour'] = df_sample['Timestamp'].dt.hour
hourly_mean = df_sample.groupby('Hour')[client_col].mean()
print('Hourly Average Load (kW):\n', hourly_mean.head(12))
```

## 2. Key Findings for NirmaanAI Engine
1. 24-hour diurnal cyclicity dominates industrial load curves.
2. Shifting non-essential machine runs out of peak tariff hours (18:00–22:00) yields immediate INR savings in Phase 14.
